**원뿔의 부피&겉넓이 계산 프로그램 6**

```python
# 원뿔 부피 및 겉넓이 계산 프로그램
# 반지름 과 높이 사용자 입력
rad = int(input("반지름을 입력하세요: "))
hei = int(input("높이를 입력하세요: "))

if rad > 0 :
    # 모선 계산
    sla = pow(rad ** 2 + hei ** 2, 0.5)
    #부피 & 겉넓이 출력
    vol = 1 / 3 * 3.14 * rad ** 2 * hei
    suf = 3.14 * rad ** 2 + 3.14 * rad * sla
    print("원뿔의 부피는", str(vol) + "입니다.")
    print("원뿔의 겉넓이는", str(suf) + "입니다.")
```

```python
#반지름, 높이 값 할당
rad = int(input("반지름 값을 입력하세요: "))
hei = int(input("높이 값을 입력하세요; "))

if rad > 0 and hei > 0 :
    #부피 출력
    print("부피의 값은 ", 1 / 3 * 3.14 * rad ** 2 * hei, "입니다.", sep="")
    #겉넓이 출력
    print("겉넓이의 값은 ", 1 / 3 * 3.14 ** 2 + 3.14 * rad * hei, "입니다.", sep="")
else :
    print("반지름과 높이의 값을 모두 양수로 입력해주세요")
```

**가장 큰 수 찾는 프로그램**

```python
# a, b, c 사용자 입력
a = int(input("첫 번째 수 입력 : "))
b = int(input("두 번째 수 입력 : "))
c = int(input("세 번째 수 입력 : "))

# a, b, c 중 가장 큰 수 출력
#max(a, b, c)
if a > b :
    if a > c :
        print(a)
    else :
        print(c)
else :
    if b > c :
        print(b)
    else :
        print(c)
```
